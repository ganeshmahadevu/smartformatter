```python
### Running Sum of 1d Array
class Solution:
    def runningSum(self, nums) :
        result = []
        a = 0
        for i in range (len(nums)):
            a = a +nums[i]
            result.append(a)
        return result
```

```python
obj = Solution()
obj.runningSum([1,8,7,4])
```

```python
### Shuffle the array
class Solution:
    def shuffle(self,nums,n):
        res=[]
        i=0
        j=n
        while j<len(nums):
            res.append(nums[i])
            res.append(nums[j])
            i+=1
            j+=1
        return res
```

```python
a = Solution()
print(a.shuffle([7,8,5,4,1,0,6,9],4))
print(a.shuffle([4,2,7,3,1,0],3))
print(a.shuffle([8,4,5,6],2))
```

```python
###  Kids With the Greatest Number of Candies
class Solution:
    def kidsWithCandies(self, candies,extraCandies):
        max= 0
        s =0
        while s < len(candies):
            if candies[s] > max:
                max = candies[s] 
            s+=1
        res = []
        i = 0
        while i < len(candies):
            total = candies[i]+ extraCandies
            if total >= max:
                res.append(True)
            else:
                res.append(False)
            i+=1
        return res
```

```python
n = Solution()
print(n.kidsWithCandies(([4,5,4,2,7]),2))
print(n.kidsWithCandies(([12,1,12]),10))
```
